```python
from pyspark.ml.classification import LogisticRegression as LR
from pyspark.ml.feature import RegexTokenizer, StopWordsRemover, CountVectorizer
from pyspark.ml.feature import OneHotEncoder, StringIndexer, VectorAssembler
from pyspark.ml.evaluation import MulticlassClassificationEvaluator
from pyspark.ml import Pipeline
from pyspark.sql.functions import col


from pyspark.sql import SparkSession

spark = SparkSession.builder \
      .appName("Crime Category Prediction") \
      .config("spark.executor.memory", "70g") \
      .config("spark.driver.memory", "50g") \
      .config("spark.memory.offHeap.enabled",True) \
      .config("spark.memory.offHeap.size","16g") \
      .getOrCreate()
# Tóm tắt: Nạp thư viện PySpark ML và tạo phiên Spark (các mức bộ nhớ chỉ là mức trần được phép dùng).
```

```python
data = spark.read.format("csv"). \
            options(header="true", inferschema="true"). \
            load("sf_crime_dataset.csv")
# Tóm tắt: Đọc file sf_crime_dataset.csv (878.049 báo cáo); dòng đầu là tên cột, Spark tự suy kiểu dữ liệu.
```

```python
data.columns
# Tóm tắt: Xem tên 9 cột của dữ liệu.
```

```python
drop_list = ['Dates', 'DayOfWeek', 'PdDistrict', 'Resolution', 'Address', 'X', 'Y']

data = data.select([column for column in data.columns if column not in drop_list])
# Tóm tắt: Bỏ 7 cột, chỉ giữ Category (nhãn cần dự đoán) và Descript (mô tả vụ việc, đầu vào).
```

```python
data.show(5)
# Tóm tắt: In 5 dòng đầu để kiểm tra dữ liệu sau khi chọn cột.
```

```python
data.printSchema()
# Tóm tắt: Xem cấu trúc bảng: 2 cột, đều kiểu chuỗi.
```

```python
# By top 20 categories
data.groupBy("Category") \
    .count() \
    .orderBy(col("count").desc()) \
    .show()
# Tóm tắt: Đếm số vụ theo từng loại tội phạm, in 20 loại nhiều nhất (thấy rõ dữ liệu mất cân bằng).
```

```python
# By top 20 descriptions
data.groupBy("Descript") \
    .count() \
    .orderBy(col("count").desc()) \
    .show()
# Tóm tắt: Đếm số lần xuất hiện của từng mô tả, in 20 mô tả phổ biến nhất.
```

```python
# regular expression tokenizer
regexTokenizer = RegexTokenizer(inputCol="Descript", outputCol="words", pattern="\\W")

# stop words
add_stopwords = ["http","https","amp","rt","t","c","the"] 

stopwordsRemover = StopWordsRemover(inputCol="words", outputCol="filtered").setStopWords(add_stopwords)

# bag of words count
countVectors = CountVectorizer(inputCol="filtered", outputCol="features", vocabSize=10000, minDF=5)
# Tóm tắt: Khai báo 3 bước xử lý văn bản: tách từ, bỏ từ dừng, đếm từ thành vector (tối đa 10.000 từ, mỗi từ xuất hiện >= 5 lần).
```

```python
label_stringIdx = StringIndexer(inputCol = "Category", outputCol = "label")
# Tóm tắt: Khai báo bước đánh số nhãn: mỗi loại tội phạm thành một số, loại phổ biến nhất = 0 (không phải one-hot).
```

```python
pipeline = Pipeline(stages=[regexTokenizer, stopwordsRemover, countVectors, label_stringIdx])

# Fit the pipeline to data.
pipelineFit = pipeline.fit(data)
dataset = pipelineFit.transform(data)
# Tóm tắt: Ghép 4 bước thành pipeline; fit học từ vựng và bảng nhãn, transform biến đổi toàn bộ dữ liệu.
```

```python
dataset.show(5)
# Tóm tắt: Xem kết quả pipeline: từ đã tách, từ đã lọc, vector đặc trưng 809 chiều, nhãn dạng số.
```

```python
### Randomly split data into training and test data sets.
(trainingData, testData) = dataset.randomSplit([0.7, 0.3], seed = 100)
print("Training Dataset Count: " + str(trainingData.count()))
print("Test Dataset Count: " + str(testData.count()))
# Tóm tắt: Chia ngẫu nhiên 70% huấn luyện / 30% kiểm tra (seed = 100) và in số dòng mỗi phần.
```

```python
# Build the model
logistic_regrssor = LR(maxIter=20, regParam=0.3, elasticNetParam=0)

# Train model with Training Data
model = logistic_regrssor.fit(trainingData)
# Tóm tắt: Huấn luyện hồi quy logistic đa lớp: tối đa 20 vòng lặp, điều chuẩn L2 với regParam = 0.3.
```

```python
# Make predictions on Test Data
predictions = model.transform(testData)

# List top crime predictions

predictions.filter(predictions['prediction'] == 0) \
    .select("Descript","Category","probability","label","prediction") \
    .orderBy("probability", ascending=False) \
    .show(n = 10, truncate = 30)
# Tóm tắt: Dự đoán trên tập kiểm tra; in 10 báo cáo được đoán là LARCENY/THEFT (nhãn 0) có xác suất cao nhất.
```

```python
evaluator = MulticlassClassificationEvaluator(predictionCol="prediction")
evaluator.evaluate(predictions)
# Tóm tắt: Đánh giá trên tập kiểm tra; chỉ số mặc định là F1 có trọng số (không phải accuracy).
```
